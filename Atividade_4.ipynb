```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import drive
drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/Colab Notebooks/Dados/air_quality_with_target.csv'
df = pd.read_csv(file_path)

df = df.replace(-200, np.nan)

def mostrar_stats(col, nome):
    print(f"\n{nome}")
    print(f"Média: {col.mean():.2f}")
    print(f"Mediana: {col.median():.2f}")
    print(f"Moda: {col.mode()[0]:.2f}")
    print(f"Variância: {col.var():.2f}")
    print(f"Desvio Padrão: {col.std():.2f}")

mostrar_stats(df['CO(GT)'], "Monóxido de Carbono (CO)")
mostrar_stats(df['NOx(GT)'], "Óxidos de Nitrogênio (NOx)")

print("\nContagem do Target")
print(df['Target'].value_counts())

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
df['Target'].value_counts().plot(kind='bar', color=['green', 'red', 'orange'])
plt.title('Distribuição da Qualidade do Ar')

plt.subplot(1, 2, 2)
plt.scatter(df['CO(GT)'], df['NOx(GT)'], alpha=0.5, c='blue')
plt.title('Relação CO vs NOx')
plt.xlabel('CO')
plt.ylabel('NOx')
plt.show()
```

**Análise Estatística**

CO (Monóxido de Carbono): A média está próxima de 0.72. Como a Mediana e a Moda são menores (0.50), isso indica que a maioria dos registros tem baixa concentração, mas existem alguns momentos de pico que elevam a média.

NOx (Óxidos de Nitrogênio): Com média de 125.46 e desvio padrão de 25.01, os dados mostram que a poluição por nitrogênio é alta e instável. O fato da Moda ser 150 indica que níveis elevados de poluição são muito frequentes.

Target (Qualidade do Ar): a soma das categorias High (325) e Moderate (308) supera a categoria Low (367). Isso significa que, na maior parte do tempo monitorado, o ar não estava em condições ideais.

**A Cidade que Respira Incerteza**

Nessa análise mostra que na maior parte do tempo, os níveis desses gases que parecem estar  sob controle (moda de $0.5$ para CO e $150$ para NOx). Mas a estabilidade é falsa. Observamos que mais de 32% dos registros são classificados como Alta poluição, significando que em cada três horas do dia, uma hora apresenta níveis de poluentes que podem ser prejudiciais à saúde.O NOx, com sua alta variância é mais imprevisível, apresentando picos repentinos que podem sobrecarregar o sistema respiratório de crianças e idosos. Já O CO, embora mais estável, mantém uma presença persistente que reduz a capacidade de transporte de oxigênio no sangue. Esses dados representam dias de maior movimento em hospitais e crises de asma que poderiam ser evitadas.
